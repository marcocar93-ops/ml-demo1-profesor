COMPARACION ENTRE TIEMPOS Y TAMAÑO ENTRE ARCHIVO CSV Y PARQUET

```python
from pathlib import Path
import pandas as pd
import time
import os

current_dir = Path.cwd()
project_root = next(
    (parent for parent in (current_dir, *current_dir.parents)
     if (parent / "data" / "raw" / "Crop_recommendation.csv").is_file()),
    None,
)
if project_root is None:
    raise FileNotFoundError("No se encontro data/raw/Crop_recommendation.csv desde la carpeta actual.")

csv_path = project_root / "data" / "raw" / "Crop_recommendation.csv"
parquet_path = project_root / "data" / "processed" / "crop.parquet"
if not parquet_path.is_file():
    raise FileNotFoundError(f"No se encontro el Parquet esperado: {parquet_path}")

csv_size = csv_path.stat().st_size / 1024
parquet_size = parquet_path.stat().st_size / 1024
print(f"CSV: {csv_size:.2f} KB")
print(f"Parquet: {parquet_size:.2f} KB")
print(f"Reduccion: {(1 - parquet_size / csv_size) * 100:.2f}%")

start = time.perf_counter()
pd.read_csv(csv_path)
csv_time = time.perf_counter() - start
start = time.perf_counter()
pd.read_parquet(parquet_path)
parquet_time = time.perf_counter() - start
print(f"Tiempo CSV: {csv_time:.4f} s")
print(f"Tiempo Parquet: {parquet_time:.4f} s")
if parquet_time > 0:
    print(f"Parquet es {csv_time / parquet_time:.2f}x mas rapido en esta medicion")
```

COMPARACION ENTRE TIEMPOS Y TAMAÑO ENTRE ARCHIVO CSV Y PARQUET

```python
import pandas as pd
import time
import os

# Tamaño de archivos
csv_size = os.path.getsize('../data/raw/crop_recommendation.csv') / 1024
parquet_size = os.path.getsize('../data/processed/crop.parquet') / 1024

print(f"CSV: {csv_size:.2f} KB")
print(f"Parquet: {parquet_size:.2f} KB")
print(f"Reducción: {(1 - parquet_size/csv_size) * 100:.2f}%")

# Tiempo de lectura
start = time.time()
pd.read_csv('../data/raw/crop_recommendation.csv')
csv_time = time.time() - start

start = time.time()
pd.read_parquet('../data/processed/crop.parquet')
parquet_time = time.time() - start

print(f"Tiempo CSV: {csv_time:.4f} s")
print(f"Tiempo Parquet: {parquet_time:.4f} s")
print(f"Parquet es {csv_time/parquet_time:.2f}x más rápido")
```
